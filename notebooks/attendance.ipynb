```python
import pandas as pd
import glob, os
```

```python
path = r"C:\Users\tlles\Documents\DA15\Capstone\baseball_payroll_wins_capstone\data"

files = glob.glob(os.path.join(path, "attendance_*.csv"))

print(len(files), "files found:", [os.path.basename(f) for f in files])
```

```python
dfs = []

for f in files:
    df = pd.read_csv(f)
    df.columns = [c.strip().lower() for c in df.columns]
    df["attendance"] = (
        df["attendance"].astype(str)
        .str.replace(",", "", regex=False)
        .str.strip()
        .replace({"":pd.NA})
        .astype("Int64")
    )
    dfs.append(df[["team","year","attendance"]])

attendace = pd.concat(dfs, ignore_index=True)
attendance.shape, attendance.head()
```

```python
TEAM_FIXES = {
    "Los Angeles Angels of Anaheim": "Los Angeles Angels",
    "Cleveland Indians": "Cleveland Guardians",   # map to your canonical name
    "Florida Marlins": "Miami Marlins",
    "Tampa Bay Devil Rays": "Tampa Bay Rays",
}

attendance["team"] = attendance["team"].str.strip().replace(TEAM_FIXES)
```
